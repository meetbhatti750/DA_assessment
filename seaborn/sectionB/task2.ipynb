```python
import pandas as pd
import numpy as np

np.random.seed(42)

restaurants = [
    "Dominos",
    "Pizza Hut",
    "McDonalds",
    "KFC",
    "Subway"
]

cities = ["Ahmedabad", "Surat", "Rajkot"]
cuisines = ["Italian", "Indian", "Fast Food"]

n = 40

data = {
    "restaurant_name": np.random.choice(restaurants, n),
    "city": np.random.choice(cities, n),
    "order_value": np.random.randint(150, 1001, n),
    "delivery_time_mins": np.random.randint(15, 51, n),
    "rating": np.round(np.random.uniform(1.0, 5.0, n), 1),
    "cuisine_type": np.random.choice(cuisines, n)
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df.to_string(index=False))

summary = (
    df.groupby("restaurant_name")
      .agg(
          mean_order_value=("order_value", "mean"),
          mean_delivery_time=("delivery_time_mins", "mean"),
          mean_rating=("rating", "mean")
      )
)

summary = summary[
    (summary["mean_rating"] > 4.0) &
    (summary["mean_delivery_time"] < 35)
]

summary = summary.sort_values(
    by="mean_order_value",
    ascending=False
)

summary = summary.reset_index()

print("\nRestaurant Performance Summary:")
print(summary.to_string(index=False))
```
