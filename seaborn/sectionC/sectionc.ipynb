```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

n = 200

restaurants = [
    "Dominos",
    "Pizza Hut",
    "McDonalds",
    "KFC",
    "Subway"
]

cities = ["Ahmedabad", "Surat", "Rajkot"]
cuisines = ["Indian", "Italian", "Fast Food"]

df = pd.DataFrame({
    "restaurant_name": np.random.choice(restaurants, n),
    "city": np.random.choice(cities, n),
    "cuisine_type": np.random.choice(cuisines, n),
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time_mins": np.random.normal(30, 8, n),
    "rating": np.random.uniform(1, 5, n),
    "discount_pct": np.random.uniform(0, 30, n)
})

for column in ["order_value", "distance_km",
               "delivery_time_mins", "rating", "discount_pct"]:

    missing_indices = np.random.choice(
        df.index,
        size=5,
        replace=False
    )

    df.loc[missing_indices, column] = np.nan

numeric_columns = df.select_dtypes(
    include="number"
).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Dataset generated and cleaned successfully!")
print("Total Rows:", len(df))
print("Missing Values Remaining:", df.isnull().sum().sum())

def summary_statistics():

    print("\n===== SUMMARY STATISTICS =====")

    for column in numeric_columns:

        values = df[column].to_numpy()

        print(f"\nColumn: {column}")
        print("Minimum:", round(np.min(values), 2))
        print("Maximum:", round(np.max(values), 2))
        print("Mean:", round(np.mean(values), 2))
        print("Standard Deviation:", round(np.std(values), 2))

    print("\nSummary statistics completed.")

def distribution_analysis():

    print("\n===== DISTRIBUTION ANALYSIS =====")

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    axes[0].hist(df["order_value"], bins=15)
    axes[0].set_title("Order Value Distribution")
    axes[0].set_xlabel("Order Value (Rs)")
    axes[0].set_ylabel("Frequency")

    axes[1].hist(df["delivery_time_mins"], bins=15)
    axes[1].set_title("Delivery Time Distribution")
    axes[1].set_xlabel("Delivery Time (Minutes)")
    axes[1].set_ylabel("Frequency")

    fig.suptitle("Food Delivery Distribution Analysis")

    plt.tight_layout()

    plt.savefig(
        "distribution_analysis.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.close(fig)

    print("Chart saved: distribution_analysis.png")

def correlation_heatmap():

    print("\n===== CORRELATION HEATMAP =====")

    correlation = df.select_dtypes(
        include="number"
    ).corr(method="pearson")

    fig, ax = plt.subplots(figsize=(10, 7))

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        vmin=-1,
        vmax=1,
        ax=ax
    )

    ax.set_title("Food Delivery Correlation Heatmap")

    plt.tight_layout()

    plt.savefig(
        "correlation_heatmap.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.close(fig)

    print(correlation.round(2))
    print("\nChart saved: correlation_heatmap.png")

def restaurant_performance():

    print("\n===== RESTAURANT PERFORMANCE REPORT =====")

    performance = (
        df.groupby("restaurant_name")
        .agg(
            total_orders=("order_value", "count"),
            mean_order_value=("order_value", "mean"),
            mean_delivery_time=("delivery_time_mins", "mean"),
            mean_rating=("rating", "mean")
        )
        .reset_index()
        .sort_values(
            by="mean_rating",
            ascending=False
        )
    )

    print(performance.round(2).to_string(index=False))

    fig, ax = plt.subplots(figsize=(9, 5))

    ax.bar(
        performance["restaurant_name"],
        performance["mean_rating"]
    )

    ax.set_title("Average Rating by Restaurant")
    ax.set_xlabel("Restaurant")
    ax.set_ylabel("Average Rating")
    ax.set_ylim(0, 5)

    plt.xticks(rotation=30)
    plt.tight_layout()

    plt.savefig(
        "restaurant_performance.png",
        dpi=150,
        bbox_inches="tight"
    )

    plt.close(fig)

    print("\nChart saved: restaurant_performance.png")

def main():

    while True:

        print("\n===================================")
        print(" FOOD DELIVERY ANALYTICS CONSOLE")
        print("===================================")
        print("1. Summary Statistics")
        print("2. Distribution Analysis")
        print("3. Correlation Heatmap")
        print("4. Restaurant Performance Report")
        print("5. Exit")

        choice = input("\nEnter your choice (1-5): ")

        if choice == "1":
            summary_statistics()

        elif choice == "2":
            distribution_analysis()

        elif choice == "3":
            correlation_heatmap()

        elif choice == "4":
            restaurant_performance()

        elif choice == "5":
            break

        else:
            print("Invalid choice! Please enter 1 to 5.")

    print("\n===================================")
    print(" FINAL EDA SUMMARY REPORT")
    print("===================================")

    # Top 3 restaurants by mean rating
    top_restaurants = (
        df.groupby("restaurant_name")["rating"]
        .mean()
        .sort_values(ascending=False)
        .head(3)
    )

    print("\nTop 3 Restaurants by Mean Rating:")

    for restaurant, rating in top_restaurants.items():
        print(f"{restaurant}: {rating:.2f}")

    correlation = df.select_dtypes(
        include="number"
    ).corr(method="pearson")

    upper_triangle = correlation.where(
        np.triu(
            np.ones(correlation.shape, dtype=bool),
            k=1
        )
    )

    correlation_pairs = upper_triangle.stack()

    if not correlation_pairs.empty:

        strongest_pair = correlation_pairs.abs().idxmax()
        strongest_value = correlation_pairs.loc[strongest_pair]

        print("\nHighest Absolute Pearson Correlation:")
        print(
            f"{strongest_pair[0]} and {strongest_pair[1]}: "
            f"{strongest_value:.3f}"
        )

    print("\nDelivery Time Statistics:")
    print(
        "Mean Delivery Time:",
        round(df["delivery_time_mins"].mean(), 2),
        "minutes"
    )

    print(
        "Standard Deviation:",
        round(df["delivery_time_mins"].std(ddof=0), 2),
        "minutes"
    )

    print("\nAll analysis completed successfully!")

    print("\nGenerated Chart Files:")
    print("1. distribution_analysis.png")
    print("2. correlation_heatmap.png")
    print("3. restaurant_performance.png")


if __name__ == "__main__":
    main()
```
