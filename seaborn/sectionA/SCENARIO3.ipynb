Question: 
Identify the most appropriate Matplotlib chart type for each of the three metrics and
justify each choice based on the nature of the data. Then explain how plt.subplots() would be
configured to arrange all three charts in a single figure, and state whether a shared x-axis would
or would not be appropriate for this layout — and why.

Answer:
Three different charts are suitable for the monthly performance report.

1. Line Chart — Monthly Orders

A line chart is appropriate for showing total orders over 12 months because it clearly displays changes and trends over time.

2. Pie Chart — Proportion of Orders by Cuisine

A pie chart is suitable for displaying the proportion of orders belonging to Fast Food, Indian, Chinese, and Desserts. Each slice represents a cuisine's share of total orders.

3. Histogram — Delivery Time Distribution

A histogram is appropriate for showing the distribution of delivery times. It helps identify common delivery times, the spread of the data, and unusually long delivery times.

Using plt.subplots():

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

orders = np.random.randint(1000, 5001, size=12)

average_order_value = np.random.uniform(200, 400, size=12)
revenue = orders * average_order_value

mean_delivery_time = 28
std_delivery_time = 4

delivery_times = np.random.normal(
    mean_delivery_time,
    std_delivery_time,
    500
)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].plot(months, orders, marker="o")

for i, value in enumerate(orders):
    axes[0].annotate(
        str(value),
        (i, value),
        textcoords="offset points",
        xytext=(0, 8),
        ha="center"
    )

axes[0].set_title("Monthly Orders")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Total Orders")
axes[0].tick_params(axis="x", rotation=45)

bar_colors = [
    "green" if value > 800000 else "red"
    for value in revenue
]

axes[1].bar(months, revenue, color=bar_colors)

axes[1].axhline(
    y=800000,
    linestyle="--",
    label="Rs 8,00,000 Threshold"
)

axes[1].set_title("Monthly Revenue")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Revenue (Rs)")
axes[1].tick_params(axis="x", rotation=45)
axes[1].legend(fontsize=8)

axes[2].hist(delivery_times, bins=15)

sample_mean = np.mean(delivery_times)

axes[2].axvline(
    sample_mean,
    linestyle="--",
    label="Mean"
)

axes[2].set_title("Delivery Time Distribution")
axes[2].set_xlabel("Delivery Time (Minutes)")
axes[2].set_ylabel("Frequency")
axes[2].legend()

fig.suptitle("Food Delivery Monthly Dashboard", fontsize=16)

plt.tight_layout()

plt.savefig(
    "food_delivery_dashboard.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print("Dashboard saved successfully!")
print("File name: food_delivery_dashboard.png")
```

Shared x-axis: A shared x-axis is not appropriate because the three charts display different types of data. The line chart uses months, the pie chart uses categories, and the histogram uses delivery time values.
