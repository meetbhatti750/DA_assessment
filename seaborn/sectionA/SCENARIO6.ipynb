Answer:
A correlation heatmap displays the Pearson correlation coefficients between numerical variables. It helps identify relationships between features and potential multicollinearity problems before training a regression model.

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation_matrix = df.corr(numeric_only=True)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

Interpretation:

A correlation of +1 indicates a perfect positive linear relationship.

A correlation of -1 indicates a perfect negative linear relationship.

A correlation near 0 indicates little or no linear relationship.

For example, a correlation of 0.90 between two features indicates a strong positive linear relationship.

Threshold for multicollinearity:

I would flag an absolute correlation greater than 0.80 as a potential multicollinearity risk.

The reason is that a strong correlation between two predictors may make regression coefficients unstable and difficult to interpret.

However, correlation alone does not prove multicollinearity. Additional checks, such as the Variance Inflation Factor (VIF), can help confirm the problem.
