Question: Explain why groupby() combined with agg() is the correct approach instead of the
for-loop strategy. Describe the specific method chain you would write to compute average order
value, average delivery time, and average rating per restaurant in a single operation, and
explain what the resulting DataFrame's index represents after the operation.

Answer:
The groupby() and agg() methods are better than a for-loop because they group data and calculate multiple statistical values efficiently in a single operation. They make the code shorter, clearer, and easier to maintain.

```python
import pandas as pd

data = {
    "restaurant_name": [
        "Dominos", "Pizza Hut", "McDonalds", "KFC", "Subway",
        "Dominos", "Pizza Hut", "McDonalds", "KFC", "Subway",
        "Dominos", "Pizza Hut", "McDonalds", "KFC", "Subway"
    ],
    "city": [
        "Ahmedabad", "Surat", "Ahmedabad", "Rajkot", "Surat",
        "Surat", "Ahmedabad", "Rajkot", "Surat", "Ahmedabad",
        "Rajkot", "Surat", "Ahmedabad", "Rajkot", "Surat"
    ],
    "order_value": [
        500, 700, 300, 450, 600,
        550, 800, 350, 400, 650,
        600, 750, 400, 500, 700
    ],
    "delivery_time_mins": [
        25, 30, 20, 28, 32,
        27, 35, 22, 30, 29,
        24, 31, 25, 32, 30
    ],
    "rating": [
        4.5, 4.2, 3.8, 4.3, 4.6,
        4.7, 4.1, 4.0, 4.5, 4.8,
        4.6, 4.4, 3.9, 4.2, 4.7
    ],
    "cuisine_type": [
        "Italian", "Italian", "Fast Food", "Fast Food", "Healthy",
        "Italian", "Italian", "Fast Food", "Fast Food", "Healthy",
        "Italian", "Italian", "Fast Food", "Fast Food", "Healthy"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

summary = (
    df.groupby("restaurant_name")
      .agg(
          mean_order_value=("order_value", "mean"),
          mean_delivery_time=("delivery_time_mins", "mean"),
          mean_rating=("rating", "mean")
      )
)

summary = summary[
    (summary["mean_rating"] > 4.0) &
    (summary["mean_delivery_time"] < 35)
]

summary = summary.sort_values(
    by="mean_order_value",
    ascending=False
)

summary = summary.reset_index()

print("\nRestaurant Performance Summary:")
print(summary.to_string(index=False))
```

Explanation:

groupby("restaurant_name") groups orders by restaurant.

agg() calculates multiple statistics for each restaurant.

mean_order_value represents the average order value.

mean_delivery_time represents the average delivery time.

mean_rating represents the average customer rating.

The resulting DataFrame uses restaurant_name as its index. Each row represents one restaurant and its calculated average values.
