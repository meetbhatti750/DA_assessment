Question: 
Describe how you would use plt.annotate() to programmatically identify and mark
the maximum order day with an arrow pointing to the peak and a text label showing the day
number and volume. Name at least two specific parameters you would pass to plt.annotate()
and explain what each one controls in the rendered chart.

Answer:
The plt.annotate() function adds text and arrows to a chart. It can highlight the day with the highest order volume.


```python
import numpy as np
import matplotlib.pyplot as plt

orders = np.array([120, 150, 180, 140, 220, 200, 250])
days = np.arange(1, len(orders) + 1)

max_index = np.argmax(orders)
max_day = days[max_index]
max_orders = orders[max_index]

plt.plot(days, orders, marker="o")

plt.annotate(
    f"Day {max_day}: {max_orders} orders",
    xy=(max_day, max_orders),
    xytext=(max_day - 2, max_orders + 30),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.title("Daily Order Volume")
plt.show()
```

Important parameters:

xy: Specifies the coordinates of the point being highlighted.

xytext: Specifies the position of the annotation text.

arrowprops: Controls the appearance of the arrow pointing to the selected point.

The np.argmax() function identifies the index of the maximum order volume, allowing the program to annotate the peak automatically.
