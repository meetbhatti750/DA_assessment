Question: 
Compare sns.boxplot() and sns.violinplot() for this specific task. Which would you
recommend and why? Provide at least two data-specific reasons that reference properties of
this dataset (10,000 rows, 3 categories, discrete 1–5 scale), and explain one situation where the
other chart type would be preferable.

Answer:
Both charts are useful for analysing customer ratings across restaurant categories.

Boxplot:

A boxplot displays the median, quartiles, spread of the data, and potential outliers.

Violin Plot:

A violin plot displays the distribution density of the data. Its width indicates where observations are more concentrated.

Recommendation: Boxplot

I would recommend a boxplot for this dataset for the following reasons:

The dataset contains 10,000 rows, and a boxplot provides a compact summary of the ratings.

The ratings use a discrete scale from 1 to 5, making it easy to compare medians, quartiles, and potential outliers across Fast Food, Fine Dining, and Casual restaurants.

It allows us to compare the three restaurant categories side by side and identify categories with unusually low or high ratings.

When is a violin plot preferable?

A violin plot is preferable when the objective is to explore the shape of the distribution and identify areas where observations are concentrated. However, because ratings are discrete, the density representation should be interpreted carefully.
