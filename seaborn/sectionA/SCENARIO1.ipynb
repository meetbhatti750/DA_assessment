Question:
Explain why NumPy vectorisation is a technically superior approach to the for-loop
suggestion. In your answer, describe what broadcasting allows when a single scalar fee rate
(e.g., Rs 5 per km) is applied to the entire array, and identify one scenario where a Python loop
would still be necessary despite NumPy being available.

Answer:
NumPy vectorisation is better than a Python for-loop because it performs operations on an entire array at once. It is faster, uses optimized numerical operations, and requires less code.

Broadcasting: Broadcasting allows NumPy to apply a single scalar value to every element of an array without manually repeating that value.


```python
import numpy as np

distances = np.array([2, 5, 8, 10])
delivery_fee = 5 * distances

print(delivery_fee)
```

A loop may be necessary when each iteration depends on the result of the previous iteration, such as processing orders sequentially when the next order depends on the previous order's result.
