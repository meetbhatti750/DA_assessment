```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

n = 200

df = pd.DataFrame({
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time": np.random.normal(30, 8, n),
    "rating": np.random.uniform(1, 5, n),
    "discount_pct": np.random.uniform(0, 30, n),
    "city": np.random.choice(
        ["Ahmedabad", "Surat", "Rajkot"], n
    )
})

print("First Five Rows:")
print(df.head())

numeric_columns = df.select_dtypes(
    include="number"
).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

sns.pairplot(
    df,
    vars=numeric_columns,
    hue="city"
)

plt.savefig("pairplot.png", dpi=200, bbox_inches="tight")
plt.close()

print("\nPairplot saved as pairplot.png")

correlation = df[numeric_columns].corr(
    method="pearson"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.savefig(
    "correlation_heatmap.png",
    dpi=200,
    bbox_inches="tight"
)

plt.close()

print("Heatmap saved as correlation_heatmap.png")
print("\nSection D completed successfully!")
```
