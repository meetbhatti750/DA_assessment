```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate synthetic dataset
n = 200

df = pd.DataFrame({
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time": np.random.normal(30, 8, n),
    "rating": np.random.uniform(1, 5, n),
    "discount_pct": np.random.uniform(0, 30, n),
    "city": np.random.choice(
        ["Ahmedabad", "Surat", "Rajkot"], n
    )
})

sns.pairplot(
    df,
    hue="city",
    diag_kind="kde"
)
plt.savefig("pairplot.png", dpi=200)
plt.close()

correlation = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)
plt.savefig("correlation_heatmap.png", dpi=200)
plt.close()
```
